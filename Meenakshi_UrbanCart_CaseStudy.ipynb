```python
#Task 1 Write a function get_product_price(product_id, products_list) that returns the Price of the matching product, or None if the ID is not found.

def get_product_price(product_id, products_list):

    for product in products_list:

        if product["Product_ID"] == product_id:

            return product["Price"]

    return None
```

```python
#Task 2 Write a function get_customer_name(customer_id, customers_list) that returns the matching customer's Name.

def get_customer_name(customer_id, customers_list):

    for customer in customers_list:

        if customer["Customer_ID"] == customer_id:

            return customer["Name"]

    return None
```

```python
#Task 3 Write a function calculate_order_value(price, quantity) that returns price * quantity.

def calculate_order_value(price, quantity):

    return price * quantity
```

```python
#Task 4 Write a function classify_order(value, high_limit=3000, low_limit=1000) that returns "High", "Medium", or "Low" using default arguments so the thresholds can be overridden later without changing the function itself.

def classify_order(value, high_limit=3000, low_limit=1000):

    if value > high_limit:
        return "High"

    elif value >= low_limit:
        return "Medium"

    else:
        return "Low"
```

```python
#Task 5 Write a function generate_invoice(order, products_list, customers_list) that uses functions 1–3 above internally and returns one neatly formatted invoice string for a single order.


# Task 5

def generate_invoice(order, products_list, customers_list):

    customer_name = get_customer_name(order["Customer_ID"], customers_list)

    price = get_product_price(order["Product_ID"], products_list)

    order_value = calculate_order_value(price, order["Quantity"])

    print("========== INVOICE ==========")
    print("Order ID      :", order["Order_ID"])
    print("Customer Name :", customer_name)
    print("Product ID    :", order["Product_ID"])
    print("Price         :", price)
    print("Quantity      :", order["Quantity"])
    print("Order Value   :", order_value)
    print("Status        :", order["Status"])
    print("=============================")

```

```python
#Task 6 Write a function calculate_total_revenue(orders_list, products_list) that loops through every order, skips cancelled ones, and returns the total revenue as a single number. Call it and confirm it matches your Module 4 answer.

def calculate_total_revenue(orders_list, products_list):

    total_revenue = 0

    for order in orders_list:

        if order["Status"] == "Cancelled":
            continue

        price = get_product_price(order["Product_ID"], products_list)

        order_value = calculate_order_value(price, order["Quantity"])

        total_revenue += order_value

    return total_revenue
```

```python
total = calculate_total_revenue(orders, products)

print("Total Revenue =", total)
```

```python
#Task 7 Write a function top_customer_by_spend(orders_list, products_list, customers_list) that returns the Name of the customer who has spent the most money (Delivered orders only).

def top_customer_by_spend(orders_list, products_list, customers_list):

    customer_spending = {}

    for order in orders_list:

        if order["Status"] != "Delivered":
            continue

        price = get_product_price(order["Product_ID"], products_list)

        order_value = calculate_order_value(price, order["Quantity"])

        customer_id = order["Customer_ID"]

        if customer_id in customer_spending:
            customer_spending[customer_id] += order_value
        else:
            customer_spending[customer_id] = order_value

    top_customer = ""
    max_spending = 0

    for customer_id in customer_spending:

        if customer_spending[customer_id] > max_spending:

            max_spending = customer_spending[customer_id]

            top_customer = customer_id

    return get_customer_name(top_customer, customers_list)
```

```python
name = top_customer_by_spend(orders, products, customers)

print("Top Customer :", name)
```

```python
#Task 8 Write a function apply_offer(price, *discounts) that accepts any number of discount percentages (e.g. 10, 5) and applies them one after another to the price, returning the final discounted price.

def apply_offer(price, *discounts):

    for discount in discounts:

        price = price - (price * discount / 100)

    return round(price, 2)
```

```python
#Task 1 Print one consolidated "UrbanCart Business Summary Report" containing:

#Total number of products
#Number of distinct categories
#Total number of customers
#Premium vs Regular customers
#Total number of orders
#Delivered/Pending/Cancelled split
#Total Revenue
##Average Order Value
#Top Customer
#Out-of-stock products
#Most popular product category



print("========== URBANCART BUSINESS SUMMARY REPORT ==========")

# Total Products
print("Total Products :", len(products))

# Distinct Categories
categories = set()

for product in products:
    categories.add(product["Category"])

print("Distinct Categories :", len(categories))

# Total Customers
print("Total Customers :", len(customers))

# Premium and Regular Customers
premium = 0
regular = 0

for customer in customers:
    if customer["Membership"] == "Premium":
        premium += 1
    else:
        regular += 1

print("Premium Customers :", premium)
print("Regular Customers :", regular)

# Total Orders
print("Total Orders :", len(orders))

# Delivered, Pending and Cancelled Orders
delivered = 0
pending = 0
cancelled = 0

for order in orders:
    if order["Status"] == "Delivered":
        delivered += 1
    elif order["Status"] == "Pending":
        pending += 1
    else:
        cancelled += 1

print("Delivered Orders :", delivered)
print("Pending Orders :", pending)
print("Cancelled Orders :", cancelled)

# Total Revenue
total_revenue = calculate_total_revenue(orders, products)
print("Total Revenue :", total_revenue)

# Average Order Value
average_order_value = total_revenue / delivered
print("Average Order Value :", round(average_order_value, 2))

# Top Customer
print("Top Customer :", top_customer_by_spend(orders, products, customers))

# Out-of-Stock Products
out_of_stock = 0

for product in products:
    if product["Stock"] == 0:
        out_of_stock += 1

print("Out-of-Stock Products :", out_of_stock)

# Most Popular Product Category
category_count = {}

for order in orders:

    if order["Status"] == "Delivered":

        for product in products:

            if order["Product_ID"] == product["Product_ID"]:

                category = product["Category"]

                if category in category_count:
                    category_count[category] += 1
                else:
                    category_count[category] = 1

                break

most_popular = ""
highest = 0

for category in category_count:

    if category_count[category] > highest:
        highest = category_count[category]
        most_popular = category

print("Most Popular Product Category :", most_popular)

print("======================================================")
```
